```python
def clean_policy_holder_age(value):
    if pd.isna(value):
        return None

    value = str(value).strip().lower()

    if "over 65" in value:
        return 66

    if "to" in value:
        parts = value.split("to")
        lower = int(parts[0].strip())
        upper = int(parts[1].strip())
        return (lower + upper) / 2

    return None


df_clean["AgeOfPolicyHolder_Clean"] = (
    df_clean["AgeOfPolicyHolder"].apply(clean_policy_holder_age)
)
```
